### Operadores Lógicos

```python
# Exibir cabeçalho da tabela
print(" A | B |  A AND B ")
print("---|---|---------")

# Gerar todas as combinações de A e B
for A in [0, 1]:
    for B in [0, 1]:
        resultado = A and B # Operação AND
        print(f" {A}|{B}|{resultado} ")
```

### Exercício 2

#### 2. Aprovado na Faculdade?

```python
n1 = float(input("Digite sua 1º nota: "))
n2 = float(input("Digite sua 2º nota: "))
n3 = float(input("Digite sua 3º nota: "))
trabalho_extra = input("Informe se fez o trabalho extra (s/n): ").lower()

soma = (n1 + n2 + n3)/3

if (soma >= 7):
    print("Você passou!")
elif (soma <= 5) or (trabalho_extra == 's'):
    print("Você passou raspando!")
else:
    print("Você repetiu.")
```

### Exercício 4

#### 4. Jogo de Adivinhação

```python
num = int(input("Digite um número de (1-10): "))
numero = 7
if (num == numero):
    print(f"Você acertou o número é {numero}")
else:
    print("Tenta novamente!")
    num = int(input("Digite um número de (1-10) - 2º chance: "))
    if (num == numero):
        print(f"Você acertou o número é {numero}")
    else:
        print("Última chance!!!")
        num = int(input("Digite um número de (1-10) - 3º chance: "))    
        if (num == numero):
            print(f"Você acertou o número é {numero}")
        else:
            print("Você perdeu.")
```

### Exercício 6

#### 6. Desconto em Compras

```python
compra = int(input("Informe quantos produtos comprou: "))
gasto = int(input("Informe quanto gastou em reais(R$): "))
cupom = input("Usou cupom na última compra (s/n)? ").lower()

if (compra >= 5 or gasto >= 100) and cupom == 'n':
    print("Você recebeu desconto em sua compra!")
else:
    print("Desculpa, sem desconto !")
```

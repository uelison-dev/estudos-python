### Desafio 4.3 - Atribuição

#### 1. Desenvolva um programa que solicite um número inteiro ao usuário. Em seguida, aplique todas as expressões de atribuição disponíveis na linguagem. Para cada operação realizada, exiba o nome da expressão e o resultado obtido.

#### Expressões de Atribuição

```python
n1 = int(input("Digite um número inteiro: "))
print(f"Valor inicial (=):{n1}")
n1 += 5
print(f"Expressão de adição (+=): {n1}")
n1 -= 3
print(f"Expressão de subtração (-=): {n1}")
n1 *= 2
print(f"Expressão de multiplicação (*=): {n1}")
n1 /= 4
print(f"Expressão de divisão (/=): {n1}")
n1 //= 3
print(f"Expressão de divisão inteira (//=): {n1}")
n1 %= 2
print(f"Expressão resto da divisão (%=): {n1}")
n1 **= 3
print(f"Expressão de exponenciação (**=): {n1}")
```

# Estruturas Condicionais

### Estrutura if

```python
idade = 18
if (idade >= 18):
    print("Você é maior de idade!")
```

### Estrutura elif

```python
idade = 16
if (idade >= 18):
    print("Você é maior de idade!")
elif (idade >= 16):
    print("Você é adolescente.")
```

### Estrutura else

```python
idade = 12
if (idade >= 18):
    print("Você é maior de idade!")
elif (idade >= 16):
    print("Você é adolescente.")
else:
    print("Você é criança")
```

### Exercício 7

#### 7. Desenvolva um programa que calcule o IMC de uma pessoa e exiba sua classificação.

```python
peso = float(input("Digite o seu peso em (kg): "))
altura = float(input("Digite a sua altura em (metros): "))
IMC = peso / (altura ** 2)
print(f"Seu IMC: {IMC:.2f}")
if (IMC <= 18.5):
    print("Abaixo do peso.")
elif(IMC <= 24.9):
    print("Peso ideal.")
elif(IMC <= 29.9):
    print("Sobrepeso")
else:
    print("Obesidade")
```

### Exercício 8

#### 8. Peça três números ao usuário e verifique se eles podem formar um triângulo.

```python
v1 = float(input("Informe o primeiro lado do triangulo: "))
v2 = float(input("Informe o segundo lado do triangulo: "))
v3 = float(input("Informe o terceiro lado do triangulo: "))

if (v1 + v2 > v3 and v1 + v3 > v2 and v2 + v3 > v1):
    print("Pode formar um triângulo.")
else:
    print("Não pode formar um triângulo.")
```

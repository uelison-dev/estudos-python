### Desafio 4.3 - Atribuição

#### 1. Desenvolva um programa que solicite um número inteiro ao usuário. Em seguida, aplique todas as expressões de atribuição disponíveis na linguagem. Para cada operação realizada, exiba o nome da expressão e o resultado obtido.

#### Expressões de Atribuição

```python
n1 = int(input("Digite um número inteiro: "))
print(f"Valor inicial (=):{n1}")
n1 += 5
print(f"Expressão de adição (+=): {n1}")
n1 -= 3
print(f"Expressão de subtração (-=): {n1}")
n1 *= 2
print(f"Expressão de multiplicação (*=): {n1}")
n1 /= 4
print(f"Expressão de divisão (/=): {n1}")
n1 //= 3
print(f"Expressão de divisão inteira (//=): {n1}")
n1 %= 2
print(f"Expressão resto da divisão (%=): {n1}")
n1 **= 3
print(f"Expressão de exponenciação (**=): {n1}")
```

## Como você faz seu sanduíche preferido ?

#### Resposta:

#### Eu pego o pão na saca, procuro uma faca de serra, após pegar a faca e com o pão na mão esquerda, eu começo a cortar, partindo o pão ao meio sem atravessar a parte de trás do pão, começo a pegar os ingredientes na geladeira após cortar o pão, começo a prepara ele passando maionese, pegando duas fatias de peito de peru, depois pegando mais duas fatias de mussarela, botando dentro do pão, fechando o mesmo e começo a comer.

## Como você faz um bolo ?

#### Resposta:

#### Para fazer um bolo, vamos precisar dos seus ingredientes, vamos preparar o ambiente para a realização do bolo, seperando por etapas cada ingrediente que será utilizado, começamos batendo alguns ingredientes na batedeira, após bater os mesmos, preparamos o tabuleiro untando ele com manteiga para que o bolo não venha grudar no mesmo, após untar derramamos o bolo com seus ingredientes já batidos dentro do tabuleiro e botamos para assar.

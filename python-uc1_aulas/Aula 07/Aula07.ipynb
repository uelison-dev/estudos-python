### Exercício 3

#### 3. Imprimir Números pares e impares

```python
for i in range(1,21):
    if (i % 2 == 0):
        print(f".:{i:02d}:. é par.")
    else:
        print(f".:{i:02d}:. é impar.")
    
```

```python
pares = []
impares = []
for i in range(1,21):
    if (i % 2 == 0):
        pares.append(i)
    else:
        impares.append(i)

print(f"Numeros pares   : {pares}")
print(f"Numeros impares : {impares}")
```

### Exercício 4

#### 4. Imprimir de 10 até 1

```python
i = 10
while i >= 0:
    print(f"{i}",end=", ")
    i -= 1
```

### Exercício 5

#### 5. Criar um programa que pergunta o nome

```python
nome = input("Digite seu nome: ")
while (nome != 'sair'):
    nome = input("Digite seu nome novamente: ")
    if (nome == 'sair'):
        print("Você saiu!")
```

### Exercício 6

#### 6. Imprimir a soma dos números

```python
soma = 0
for i in range(1, 101):
    soma = soma + i
print(f"O valor da soma de todos os numeros de 1 até 100 é {soma}")
```

### Exercício 7

#### 7. Calcular o fatorial de um número usando while

```python
i = int(input("Digite um número: "))
r = 1
n_aux = i
while n_aux >= 1:
    r = r * n_aux
    n_aux = n_aux - 1
print(f"O fatorial de .:{i}:. é .:{r}:.")
    
```

### 8.1 Funções simples

```python
def soma (a, b):
    soma = a + b
    return soma
```

```python
resultado = soma(12.4, 15)
print (f"O resultado é .:{resultado:4.2f}:.")
```

```python
def soma_2 (n1, n2):
    soma = n1+n2
    print(soma)
```

```python
soma_2(12.4,15)
```

```python
resposta = soma_2(12,50)
```

```python
def potencia(numero, expoente):
    resultado = numero ** expoente
    return resultado
```

```python
n1 = 2
exp = 8

print(potencia(expoente=exp, numero=n1))
```

```python
n2 = 5
exp = 4
resultado = potencia(n2, exp)
resultado2 = potencia(exp, n2)
```

```python
print(f"{resultado} <> {resultado2}")
```

# Passagem de argumentos com valor "Default"

```python
def saudacao(display="sem nome"):
    print(f"Olá, {display}!!!")
```

```python
saudacao("Uelison")
```

```python
saudacao()
```

# Passagem de argumentos arbitrários

```python
def saudacoes(*nomes):
    #print(nomes)
    for nome in nomes:
        saudacao(nome)
```

```python
saudacoes ("Hugo", "Matheus", "Tadeus")
```

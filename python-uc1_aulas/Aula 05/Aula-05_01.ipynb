### Exercício 1

```python
n1 = float(input("Digite o 1º número : "))
n2 = float(input("Digite o 2º número : "))

soma = n1 + n2
sub = n1 - n2
multi = n1 * n2
div = n1 / n2

print(f"A soma entre {n1} e {n2} é: {soma}")
print(f"A subtração entre {n1} e {n2} é: {sub}")
print(f"A multiplicação entre {n1} e {n2} é: {multi}")
print(f"A divisão entre {n1} e {n2} é: {div}")
```

### Exercício 8

```python
capital = float(input("Digite o valor inical : "))
taxa_percentual = float(input("Digite o valor da taxa de juros : "))
tempo = float(input("Digite o tempo em meses (1-12) :"))
i = taxa_percentual / 100
M = capital * (1 + i) ** tempo
print(f"Montante final do investimento é: R${M:.2f}")
```

### Exercício 7

```python
peso = float(input("Digite o seu peso e KG : "))
alt = float(input("Digite a sua altura em metros : "))
IMC = peso/(alt**2)
print(f"O seu IMC é: {IMC:.2f}")
if (IMC < 18.5):
    print(".:Classificação:. Abaixo do peso")
elif (18.5 <= IMC <= 24.9):
    print(".:Classificação:. Peso ideal")
elif (25 <= IMC <= 29.9):
    print(".:Classificação:. Sobrepeso")
else:
    print(".:Classificação:. Obesidade!")

```

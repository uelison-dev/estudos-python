```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    ["A", "B", "C"],
    [4, "D", 0.77]
]

print(matriz)

for linha in matriz:
    for elemento in linha:
        print(elemento, end=" ")
    print()
```

```python
for linha in matriz:
    print("[",end="")
    for elemento in linha:
        print(f"|{elemento}|,", end="\t")
    print("]",end="")
    print()
print("Fim !")
```

```python
user = "Uelison"
password = 777

user = (input("Digite o seu nome de usuário: "))
password = float(input("Digite a sua senha: "))
idade = int(input("Digite a sua idade: "))

if (user == "Uelison" and password == 777):
    if (idade >= 18):
        print(f"{user} logou com sucesso e você é maior de 18 anos!")
    else:
        print("Você não possui idade, porém acertou o usuário e senha!")
elif (user == "Uelison" or password == 777):
    if (idade < 18):
        print("Você acertou o usuário ou senha!")
    else:
        print("Porém você não possui idade minima")
else:
    print("Usuário e senha incorreto!")
```

### Exercício 2

```python
nota1 = float(input("Digite a 1º nota: "))
nota2 = float(input("Digite a 2º nota: "))
nota3 = float(input("Digite a 3º nota: "))
trabalho_extra = input("Você fez o trabalho extra: (S/N) ?").lower()

media = (nota1 + nota2 + nota3)/3

if (media >= 7 and trabalho_extra == 'n'):
    print("Você passou de série!")
elif (media < 7 and trabalho_extra == 's'):
    print("Você passou por pouco!")
else:
    print("Você não possui a média e não fez o trabalho")
```

```python
import random
segredo = random.randint(0,10)
tentativas = 0
while tentativas < 3:
    numero = int(input("Digite um numero: "))
    if numero == segredo:
        print("Parabéns! Você acertou!")
        break
    else:
        print("Você errou, tente novamente")
        tentativas+=1
print(f"O número secreto era .:{segredo}:.")
```

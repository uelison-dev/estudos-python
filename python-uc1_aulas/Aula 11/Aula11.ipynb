## Usando o método fromkeys()

```python
chaves = ["nome", "dt_nascimento", "cep", "logradouro", "bairro", "cidade", "estado"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)

# Resultado: {'nome': 'N/A, 'idade': 'N/A', 'cidade': 'N/A'}
print(usuario)
```

```python
usuario["nome"]=input("Digite seu nome: ")
usuario["dt_nascimento"]=input("Digite sua data de nascimento (aa/bb/cccc): ")
usuario["cep"]=input("Digite seu cep:")
usuario["logradouro"]=input("Digite seu logradouro: ")
usuario["bairro"]=input("Digite seu bairro: ")
usuario["cidade"]=input("Digite sua cidade: ")
usuario["estado"]=input("Digite seu estado: ")
```

```python
print(usuario["nome"])
print(usuario["dt_nascimento"])
print(usuario["cep"])
print(usuario["logradouro"])
print(usuario["bairro"])
print(usuario["cidade"])
print(usuario["estado"])
```

```python
for chave, valor in usuario.items():
    print(f"{chave.capitalize():_<20}: {valor.capitalize()}")
```

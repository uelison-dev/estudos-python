### Exercício 3: Função par_ou_impar

```python
def par_ou_impar(n):
    if n % 2 == 0:
        return "Par"
    else:
        return "Ímpar"
# Exemplo de chamada
print("O número 4 é", par_ou_impar(4))
print("O número 7 é", par_ou_impar(7))
```

### Exercício 4: Função fatorial

```python
def fatorial(n):
    if (n < 0):
        return "Número inválido para fatorial"
    elif (n == 0):
        return 1
    else:
        resultado = 1
        for i in range(1, n + 1):
            resultado *= i
        return resultado
print("Fatorial de 5:", fatorial(5))
```

### Exercício 5: Função maior_elemento

### Exercício 7: Função eh_palindromo

```python
def eh_palindromo(string):
    return string == string[::-1] # Espelho
print("ana é palindromo?", eh_palindromo("ana"))
print("python é palindromo?", eh_palindromo("python"))
```

### Exercício 8: Função contar_vogais

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    contador = 0
    for char in texto:
        if char in vogais:
            contador += 1
    return contador

print("Número de vogais em 'Python':", contar_vogais("Python"))
```

### Exercício 9: Função tabuada

```python
def tabuada(n):
    resultado = []
    for i in range(1, 11):
        resultado.append(n * i)
    return resultado
# Exemplo de chamada
print("Tabuada do 5:", tabuada(5))
```

### 🐍 Criando uma Classe em Python

```python
# Definindo uma classe chamada 'Carro'
class Carro:
    # Método construtor (__init__) para inicializar os atributos
    def __init__(self, marca, modelo, portas = 4, ano="2026", cor = "Branca"):
        self.marca = marca      # Atributo
        self.modelo = modelo    # Atributo
        self.ano = ano          # Atributo
        self.cor = cor
        self.potencia = "1.0"       
        self.tp_cambio = "auto"  
        self.qt_portas = portas
        self.dir_hidraulica = True 
        self.tp_tracao = "Fwd"
        self.cambio_automatico = False         
        self._4p4 = False           
        self.ligado = False     # Atributo com valor padrão


    # Método para ligar o carro
    def ligar(self):
        if not self.__ligado:
            self.__ligado = True
            print(f"{self.marca} {self.modelo} está ligado.")
        else:
            print(f"{self.marca} {self.modelo} já está ligado.")

    # Método para desligar o carro
    def desligar(self):
        if self.__ligado:
            self.__ligado = False
            print(f"{self.marca} {self.modelo} está desligado.")
        else:
            print(f"{self.marca} {self.modelo} já está desligado.")

    # Método para exibir informações do carro
    def exibir_info(self):
        status = "ligado" if self.ligado else "desligado"
        self.m_cor()
        self.m_potencia()
        self.m_tp_tracao()
        self.m_cambio_automatico()
        print(f"{self.marca} {self.modelo} ({self.ano}) está {status}.")

    def mudar_cor(self, cor):
        self.cor = cor
        return True

    def mudar_cambio(self, cambio):
        cambio = cambio.lower()
        if cambio == "auto":
            self.tp_cambio = "auto"
        elif cambio == "manual" :
            self.tp_cambio = "manual"
        else:
            return False
        return True

    def mudar_tracao(self):
        if self.tp_tracao == 2:
            self.tp_tracao = 4
        else:
            self.tp_tracao = 2
        return True

    def mudar_potencia(self, pot="1.0"):
        try:
            v1_pot=float(pot)
            self.potencia=v1_pot
            return True
        except:
            return False
```

```python
# Criando um objeto da classe Carro
meu_carro = Carro("Toyota", "Corolla", 2020)
meu_carro2 = Carro ("Fiat", "Palio")
```

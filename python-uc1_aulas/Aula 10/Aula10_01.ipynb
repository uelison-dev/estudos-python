# 1.2 Matrizes

```python
# Criando uma matriz 3x3
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

```python
import random
QT_LINHAS = 4
QT_COLUNAS = 5
matriz = []
for _LINHA in range (QT_LINHAS):
    linha = []
    for _COLUNA in range (QT_COLUNAS):
        linha.append(random.randint(1,100))
    matriz.append(linha)

print(matriz)
```

```python
print("Elemento (0,0):", matriz[0][0]) # Saída 1
print("Elemento (2,1):", matriz[2][1]) # Saída 8
```

```python
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:03d}", end=' ')
    print()
```

# 10.1.1 Criando e Exibindo uma Matriz

```python
matriz = []
for i in range(4):
    linha = []
    for j in range(4):
        valor = int(input(f"Digite o valor para [{i}][{j}]: "))
        linha.append(valor)
    matriz.append(linha)

for linha in matriz:
    print(linha)
```

```python
import random

def criar_matriz(qt_linhas=4, qt_colunas=4):
    matriz = []
    for _LINHA in range (qt_linhas):
        linha = []
        for _COLUNA in range (qt_colunas):
            linha.append(random.randint(1,100))
        matriz.append(linha)
    return (matriz)

matriz = criar_matriz(5, 5)

for linha in matriz:
    print(linha)
```

```python
def imprime_matriz(matriz=[]):
    for linha in matriz:
        for elemento in linha:
            print(f"{elemento:4d}", end=" ")
        print()
```

```python
minha_matriz = criar_matriz(6, 7)
```

```python
imprime_matriz(minha_matriz)
```

```python
def imprime_matriz_v2(matriz=[]):
    if isinstance(matriz,list):    
        for linha in matriz: 
            if isinstance(linha, list):
                for elemento in linha :
                    print(f"{elemento:4d}",end=" ")
                print()
    else:
        print("[ERRO]")
```

# 2. Criando Dicionários

```python
# Dicionário vazio
d1 = {}
d2 = dict()

# Dicionário com elementos
telefones = {"Alice": "1234-5678", "Bob": "8765-4321"}
print(telefones)
```

```python
# Usando pares chave-valor
d3 = dict(nome="Carlos", idade =25)
print(d3)

# A partir de uma lista de tuplas
lista_tuplas = [{"a", 1}, ("b", 2)]
d4 = dict(lista_tuplas)
print(d4)
```

```python
chaves = ["nome", "idade", "cidade"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)

# Resultado: {'nome': 'N/A, 'idade': 'N/A', 'cidade': 'N/A'}
print(usuario)

usuario["nome"] = "Luis Rodrigo"
usuario["idade"] = 50
usuario["cidade"] = "Petrópolis"
print(usuario)
```

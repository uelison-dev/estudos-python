### Exercício 1

#### 1. Calculadora Básica:

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

soma = (numero1+numero2)
subtracao = (numero1-numero2)
multiplicacao = (numero1*numero2)
divisao = (numero1/numero2)

print(f"A soma é: {soma}")
print(f"A subtração é: {subtracao}")
print(f"A multiplicação é: {multiplicacao}")
print(f"A divisão é: {divisao}")
```

### Exercício 5

#### 5. Calculadora de IMC:

```python
peso = float(input("Digite o seu peso em (KG): "))
altura = float(input("Digite a sua altura em (metros): "))
IMC = peso / altura ** 2
print(f"O seu IMC é: {IMC:.2f}")
```

### Exercício 8

### Exercício 9

#### 9. Bhaskara:

```python
# 1. Entrada de dados: pede os valores ao usuário
a = float(input("Digite o valor de a: "))
b = float(input("Digite o valor de b: "))
c = float(input("Digite o valor de c: "))

# O valor de 'a' não pode ser zero
if (a == 0):
    print("O valorde 'a' não pode ser zero em uma equação do segundo grau")
else:
    # 2. Cálculo de Delta (Δ = b² - 4ac)
    delta = (b ** 2) - (4 *a * c)
    print(f"O valor de delta é: {delta}")

    # 3. Verificação das raízes com base no Delta
    if (delta < 0):
        print("A equação não possui raízes reais.")
    elif (delta == 0):
        x = -b / (2 * a)
        print(f"A equação possui uma raiz: x = {x}")
    else:
        x1 = (-b + (delta ** 0.5)) / (2 * a)
        x2 = (-b - (delta ** 0.5)) / (2 * a)

        print("A Equação possui duas raízes reais:")
        print(f"x1 = {x1}")
        print(f"x2 = {x2}")
```

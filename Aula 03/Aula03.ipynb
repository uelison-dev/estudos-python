### Exercício 1

#### 1. Calculadora Básica:

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

soma = (numero1+numero2)
subtracao = (numero1-numero2)
multiplicacao = (numero1*numero2)
divisao = (numero1/numero2)

print(f"A soma é: {soma}")
print(f"A subtração é: {subtracao}")
print(f"A multiplicação é: {multiplicacao}")
print(f"A divisão é: {divisao}")
```

### Exercício 2

#### 2. Média de notas:

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = (nota1 + nota2 + nota3)/3

print(f"A média total das notas foi dê: {media}")
```

### Exercício 3

#### 3. Conversor de temperatura:

```python
temperatura = float(input("Digite a temperatura para conversão: "))
F = (temperatura * 9/5) + 32
print(f"A temperatura em Celsius é: {temperatura} e em Fahrenheit é: {F}")
```

### Exercício 4

#### 4. Área de um círculo:

```python
r = float(input("Digite o raio do circulo: "))
A = 3.14 * r ** 2
print(f"O área do circulo é: {A}")
```

### Exercício 5

#### 5. Calculadora de IMC:

```python
peso = float(input("Digite o seu peso em (KG): "))
altura = float(input("Digite a sua altura em (metros): "))
IMC = peso / altura ** 2
print(f"O seu IMC é: {IMC:.2f}")
```
